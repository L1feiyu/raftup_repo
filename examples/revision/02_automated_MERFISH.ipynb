# MERFISH automatic cutoff selection
Use prepared slices and their cached feature-cost matrix to reproduce a pair. Labels are not selection inputs. For all benchmark pairs, repeat this call for each manifest.


## Data and local paths
Data are not uploaded to GitHub. Download sources and preparation instructions: [DATA.md](DATA.md).
Default prepared-input folder: `data/revision/merfish_0409/` in the repository; set `RAFTUP_TUTORIAL_DATA` to override it. Relative paths inside `pair.json` resolve against that file. Outputs go to `outputs/revision/02_automated_MERFISH/`; set `RAFTUP_OUTPUT_ROOT` to override. The real-data notebooks do not substitute synthetic data when inputs are missing.


```python
from pathlib import Path
import os, json
import numpy as np
import raftup
from raftup import run_automated_alignment, run_manual_alignment
from raftup.experiments import load_pair, save_alignment, coverage
REPO = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').is_file()), Path.cwd())
DATA = Path(os.environ.get('RAFTUP_TUTORIAL_DATA', str(REPO / 'data/revision/merfish_0409'))).expanduser().resolve()
OUT = Path(os.environ.get('RAFTUP_OUTPUT_ROOT', str(REPO / 'outputs/revision/02_automated_MERFISH'))).expanduser().resolve()
print('Inputs:', DATA, '\nOutputs:', OUT)
print('RAFT-UP:', raftup.__version__, raftup.__file__)
```

```python
# Set DATA or RAFTUP_TUTORIAL_DATA to your prepared input directory.
# pair.json contains file paths, exact cost-matrix cell order and spatial calibration.
a, b, cost, cfg = load_pair(DATA / 'pair.json')
print(a.shape, b.shape, cost.shape)
```

```python
P, decision = run_automated_alignment(a, b, cost,
    technology='merfish', pixels_per_100um=cfg['pixels_per_100um'])
print(decision)
print(coverage(P))
save_alignment(P, decision, OUT / 'merfish' / cfg['name'])
```
